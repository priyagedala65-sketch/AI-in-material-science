```python
import seaborn as sns
```

```python
# 1. Load the cleaned dataset containing family classifications
df = pd.read_csv('cleaned_cp_dataset.csv')
```

```python
print(df.columns.tolist())
print(df.head())
```

```python
# Create chemical family from the formula
def get_family(formula):
    formula = str(formula)

    if "O" in formula:
        return "Oxides"
    elif "S" in formula:
        return "Sulfides"
    elif "N" in formula:
        return "Nitrides"
    elif any(x in formula for x in ["F", "Cl", "Br", "I"]):
        return "Halides"
    else:
        return "Other"


df["family"] = df["formula"].apply(get_family)

# 2. Create a FacetGrid for Cp vs T across chemical families
g = sns.FacetGrid(
    df,
    col="family",
    hue="formula",
    height=5,
    aspect=1.2,
    palette="tab10"
)

g.map(
    plt.plot,
    "T",
    "Cp",
    marker="o",
    linewidth=2
)

g.add_legend(title="Material Formula")

plt.show()
# Ensure style settings for publication quality
sns.set_theme(style="whitegrid", font_scale=1.1)
```
